```python
from pyspark.sql.window import Window
from pyspark.sql.functions import (
    col,
    sum,
    round
)

# Jointure orders + customers

# CA par client
df_customer_revenue = df_orders_customers.groupBy(
    "customer_unique_id"
).agg(
    round(
        sum("price"),
        2
    ).alias("customer_revenue")
)

# Total global
total_revenue = df_customer_revenue.agg(
    sum("customer_revenue").alias("total_revenue")
).collect()[0]["total_revenue"]

# Fenêtre cumulative
window_spec = Window.orderBy(
    col("customer_revenue").desc()
)

# CA cumulé
df_customer_revenue = df_customer_revenue.withColumn(
    "cumulative_revenue",
    sum("customer_revenue").over(window_spec)
)

# Pourcentage cumulé
df_customer_revenue = df_customer_revenue.withColumn(
    "cumulative_percentage",
    round(
        (col("cumulative_revenue") / total_revenue) * 100,
        2
    )
)

# Clients représentant 80%
df_top_80 = df_customer_revenue.filter(
    col("cumulative_percentage") <= 80
)

df_top_80.show(truncate=False)

```

```python
from pyspark.sql.functions import (
    col,
    date_format,
    sum
)

# Colonne mois
df_orders_month = order_df.withColumn(
    "order_month",
    date_format(
        col("order_purchase_timestamp"),
        "yyyy-MM"
    )
)

# Tableau croisé
df_pivot = df_orders_month.groupBy("order_month") \
    .pivot("seller_state") \
    .agg(
        round(sum("price"), 2)
    ) \
    .orderBy("order_month")

df_pivot.display(truncate=False)

```

```python
from pyspark.sql.functions import (
    col,
    date_format,
    countDistinct,
    sum,
    round
)

# Colonne mois
df_orders_month = order_df.withColumn(
    "order_month",
    date_format(
        col("order_purchase_timestamp"),
        "yyyy-MM"
    )
)

# Statistiques mensuelles
df_monthly_stats = df_orders_month.groupBy(
    "order_month"
).agg(
    countDistinct("order_id").alias("nb_orders"),
    round(sum("price"), 2).alias("total_amount")
).orderBy("order_month")

df_monthly_stats.show(truncate=False)

```

```python
from pyspark.sql.functions import (
    avg,
    stddev,
    col
)

# Statistiques globales
stats = order_df.agg(
    avg("price").alias("avg_price"),
    stddev("price").alias("std_price")
).collect()[0]

avg_price = stats["avg_price"]
std_price = stats["std_price"]

# Seuil
threshold = avg_price + (3 * std_price)

# Outliers
df_outliers = order_df.filter(
    col("price") > threshold
)

df_outliers.select(
    "order_id",
    "product_id",
    "price"
).show(truncate=False)

spark.catalog.clearCache()
```

```python
from pyspark.sql.functions import (
    collect_set
)

# Groupement produits
df_products_array = order_df.groupBy(
    "order_id"
).agg(
    collect_set("product_id").alias("products_array")
)

df_products_array.show(truncate=False)

spark.catalog.clearCache()
```
